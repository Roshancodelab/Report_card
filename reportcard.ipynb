```python
main_db={}
n=int(input("ENTER THE NUMBER OF RECOD OF STUDENT:"))

for i in range (n):
     print(f"ENTER THE DETAIL OF STUDENT {i+1}:")
     key_name=input("ENTER THE ID OF STUDENT :")
     name=input("ENTER THE NAME OF STUDENT :")
 

     subject_marks={}
     s=int(input("ENTER THE NUMBER OF SUBJECT:"))
     for j in range(s):
          subject=input(f"ENTER THE NAME OF SUBJECT {j+1} :")
          score=int(input(f"ENTER THE MARKS OF {subject}:"))
          subject_marks[subject]=score
          main_db[key_name]={
               "Student_name":name,
               "marks":subject_marks
          }

print("-"*10,"STUDENT REPORT CARD","-"*10)
for std_id,std_info in main_db.items():
     print(f" [ID : {std_id}] Student name : {std_info['Student_name']}")
     print("-"*35)
     print(f"{'Subject':<15} | {'Marks':<15} ")
     print("-"*35)

     inner_marks=std_info["marks"]
     for subject_name,score_value in inner_marks.items():
          print(f"{subject_name:<15} | {score_value:<15}")

     total=sum(inner_marks.values())  
     average= total / len(inner_marks)

     print("-"*35)
     print(f"TOTAL MARKS : {total}/{len(inner_marks)*100}")
     print(f"PERCENTEGE  : {average:.2f}%")
     print("-"*35)   


     
```
